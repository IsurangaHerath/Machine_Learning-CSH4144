```python
df=pd.read_csv("diabetes.csv")

print("Before clean")
details1={
    "shape":df.shape,
    "NUll values\n": df.isnull().sum()
}
print(details1)

invalid_zero_columns = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI"
]

df[invalid_zero_columns]=df[invalid_zero_columns].replace(0,np.nan)
print("After replace")
details2={
    "shape":df.shape,
    "NUll values\n": df.isnull().sum()
}
print(details2)

for i in invalid_zero_columns:
    df[i]=df[i].fillna(df[i].median())

print("After clean")
details2={
    "shape":df.shape,
    "NUll values\n": df.isnull().sum()
}
print(details2)
```
