```python
def add(x,y):
    return x+y
def sub(x,y):
    return x-y
def mul(x,y):
    return x*y
def div(x,y):
    return x/y

print("Addition: ",add(5,4))
print("Substraction: ",sub(4,2))
print("Multiplication: ",mul(4,6))
print("Division: ",div(6,3))
```
